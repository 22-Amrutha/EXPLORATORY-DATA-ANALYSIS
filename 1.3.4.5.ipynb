```python
#14
import numpy as np

a = np.random.randint(1, 10, (2, 3))
b = np.random.randint(1, 10, (2, 3))
print(a)
print(b)

# Vertical concatenation
v1 = np.vstack((a, b))
v2 = np.concatenate((a, b), axis=0)
print(v1)
print(v2)

print("Vertical results identical:", np.array_equal(v1, v2))

# Horizontal concatenation
h1 = np.hstack((a, b))
h2 = np.concatenate((a, b), axis=1)
print(h1)
print(h2)

print("Horizontal results identical:", np.array_equal(h1, h2))
```

```python
#21
import numpy as np

student_names = np.array(["Anu", "Ravi", "Priya", "Kiran", "Meena"])
scores = np.array([80, 65, 92, 70, 78])

# Get sorting indices
i = np.argsort(scores)

print("Students ranked from lowest to highest score:")

for index in i:
    print(student_names[index], scores[index])
```

```python
#22
import numpy as np

X = np.random.rand(10, 2)
print(X)

dist_sq = np.sum( (X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=2)

indices = np.argsort(dist_sq, axis=1)

print("\n2 nearest neighbors for each point:")

for i in range(10):
    neighbors = indices[i, 1:3]
    print("Point", i, "->", neighbors)
```

```python
#24
import numpy as np

dtype = [('title', 'U30'),('author', 'U20'),('price', 'f8')]
books = np.array([ ('Python Basics', 'John Smith', 450.0),('Data Science', 'Anita Rao', 550.0), ('Machine Learning', 'Rahul Kumar', 650.0), ('NumPy Guide', 'Priya Sharma', 350.0)], dtype=dtype)
# (a) All book titles
print("Titles:", books['title'])

# (b) Complete record of the second book
print("Second book:", books[1])

# (c) Titles of books priced below 500
print("Books below 500:", books['title'][books['price'] < 500])
```

```python
#25
import numpy as np

# Dictionary form
dtype_dict = np.dtype({
    'names': ('title', 'author', 'price'),
    'formats': ('U30', 'U20', 'f8')
})

# List-of-tuples form
dtype_list = np.dtype([
    ('title', 'U30'),
    ('author', 'U20'),
    ('price', 'f8')
])

print("Equivalent?", dtype_dict == dtype_list)
```

```python
#26
# Convert to recarray
books_rec = books.view(np.recarray)

# Access using attribute-style
print("Prices:", books_rec.price)
```

```python
#28
# Define dtype
student_dtype = np.dtype([
    ('roll_no', 'i4'),
    ('name', 'U20'),
    ('marks', 'f8')
])

# Create structured array
students = np.array([
    (101, "ashish", 85.5),
    (102, "Bhoomi", 92.0),
    (103, "Chim", 78.0),
    (104, "David", 88.5),
    (105, "laxmi", 95.0)
], dtype=student_dtype)

# Sort by marks (descending)
sorted_students = np.sort(students, order='marks')[::-1]

print(sorted_students)
```
