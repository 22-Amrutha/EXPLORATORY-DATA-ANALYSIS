```python
#31
area = pd.Series([60330, 342239, 307713, 191791, 308350],
                 index=["Kerala", "Maharashtra", "Uttar Pradesh", "Tamil Nadu", "Rajasthan"])
population = pd.Series([35_000_000, 123_000_000, 231_000_000, 78_000_000, 82_000_000],
                       index=["Kerala", "Maharashtra", "Uttar Pradesh", "Tamil Nadu", "Rajasthan"])

states = pd.DataFrame({"area": area, "population": population})
states["density"] = states["population"] / states["area"]

print(states)
```
